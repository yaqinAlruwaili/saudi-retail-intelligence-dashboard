Section 2 — Cleaning process

```python
#01 — Remove duplicate rows
df = df.drop_duplicates()
```

```python
#02 — Remove cancelled invoices
df = df[
    ~df["InvoiceNo"].astype(str).str.startswith("C")
]
```

```python
#03 — Remove rows with missing descriptions
df = df.dropna(subset=["Description"])
```

```python
#04 — Remove zero or negative prices
df = df[df["UnitPrice"] > 0]
```

```python
#05 — Remove zero or negative quantities
df = df[df["Quantity"] > 0]
```

```python
#06 — Reset the index
df = df.reset_index(drop=True)
```

```python
#07 — Validate the cleaned data
print("=" * 50)
print("AFTER CLEANING")
print("=" * 50)

print(f"Rows: {df.shape[0]:,}")

print(f"Duplicate Rows: {df.duplicated().sum()}")

print(f"Missing Description: {df['Description'].isna().sum()}")

print(f"Invalid Quantity: {(df['Quantity'] <= 0).sum()}")

print(f"Invalid Unit Price: {(df['UnitPrice'] <= 0).sum()}")

print(f"Cancelled Orders: {df['InvoiceNo'].astype(str).str.startswith('C').sum()}")

print(f"Missing CustomerID: {df['CustomerID'].isna().sum()}")
```

```python
#08 — Save the cleaned dataset.
df.to_csv(
    "../data/processed/clean_transactions.csv",
    index=False
)
```
