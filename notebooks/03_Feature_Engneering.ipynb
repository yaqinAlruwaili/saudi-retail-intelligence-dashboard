Section 1 — Sales Features

```python
#01 — Load the Library
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
#02 — Load the Clean Dataset
df = pd.read_csv(
    "../data/processed/clean_transactions.csv",
    parse_dates=["InvoiceDate"]
)

df.head()
```

```python
#03 — Create Sales Amount
df["SalesAmount"] = (
    df["Quantity"] *
    df["UnitPrice"]
).round(2)
```

```python
#Verify
df[
    ["Quantity",
     "UnitPrice",
     "SalesAmount"]
].head()
```

```python
#04 — Generate Product Cost
np.random.seed(42)

unique_products = df["StockCode"].unique()

cost_ratios = pd.DataFrame({
    "StockCode": unique_products,
    "CostRatio": np.random.uniform(
        0.50,
        0.80,
        len(unique_products)
    )
})
```

```python
#05 — Merge it back
df = df.merge(
    cost_ratios,
    on="StockCode",
    how="left"
)
```

```python
#06 — Calculate Unit Cost
df["UnitCost"] = (
    df["UnitPrice"] *
    df["CostRatio"]
).round(2)
```

```python
#07 — Calculate Total Cost
df["TotalCost"] = (
    df["Quantity"] *
    df["UnitCost"]
).round(2)
```

```python
#08 — Calculate Profit
df["Profit"] = (
    df["SalesAmount"] -
    df["TotalCost"]
).round(2)
```

```python
#09 — Calculate Profit Margin % 
df["ProfitMargin"] = np.where(
    df["SalesAmount"] > 0,
    df["Profit"] / df["SalesAmount"],
    0
)
```

```python
#10 — Convert to percentage
df["ProfitMargin"] = (
    df["ProfitMargin"] * 100
).round(2)
```

```python
#11 — Validate the Features
df[
    [
        "SalesAmount",
        "UnitCost",
        "TotalCost",
        "Profit",
        "ProfitMargin"
    ]
].describe()
```

```python
#12 — Inspect Sample Rows
df[
    [
        "Description",
        "Quantity",
        "UnitPrice",
        "SalesAmount",
        "UnitCost",
        "TotalCost",
        "Profit",
        "ProfitMargin"
    ]
].head(15)
```

```python
#13 — Save the Enriched Dataset
df.to_csv(
    "../data/processed/transactions_enriched.csv",
    index=False
)
```

Section 2 — Date Features

```python
#01 — Create Invoice Date Only
df["InvoiceDateOnly"] = df["InvoiceDate"].dt.date
```

```python
#02 — Year
df["Year"] = df["InvoiceDate"].dt.year
```

```python
#03 — Quarter
df["Quarter"] = "Q" + df["InvoiceDate"].dt.quarter.astype(str)
```

```python
#04 — Month Number
df["Month"]=df["InvoiceDate"].dt.month
```

```python
#05 — Month Name
df["MonthName"] = df["InvoiceDate"].dt.month_name()
```

```python
#06 — Month Abbreviation
df["MonthShort"] = df["InvoiceDate"].dt.strftime("%b")
```

```python
#07 — Week Number
df["Week"] = df["InvoiceDate"].dt.isocalendar().week
```

```python
#08 — Day of Month
df["Day"] = df["InvoiceDate"].dt.day
```

```python
#09 — Day Name 
df["DayName"] = df["InvoiceDate"].dt.day_name()
```

```python
#10 — Day Number
df["DayOfWeek"] = df["InvoiceDate"].dt.dayofweek
```

```python
#11 — Weekend Flag
df["IsWeekend"] = df["DayOfWeek"] >= 5
```

```python
#12 — Season

#Create a function
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

#apply it
df["Season"] = df["Month"].apply(get_season)
```

```python
#13 — Simple improvment (YearMonth)
df["YearMonth"] = (
    df["InvoiceDate"].dt.strftime("%Y-%m")
)
```

```python
#14 — Validation
df[
    [
        "InvoiceDate",
        "InvoiceDateOnly",
        "Year",
        "Quarter",
        "Month",
        "MonthName",
        "Week",
        "Day",
        "DayName",
        "IsWeekend",
        "Season"
    ]
].head(15)
```

```python
#15 — Save progress
df.to_csv(
    "../data/processed/transactions_enriched.csv",
    index=False
)
```

Section 3 — Business Rule Features

```python
#01 — DiscountPercent
np.random.seed(42)

discounts = [0, 5, 10, 15, 20]

probabilities = [
    0.70,
    0.15,
    0.08,
    0.05,
    0.02
]

df["DiscountPercent"] = np.random.choice(
    discounts,
    size=len(df),
    p=probabilities
)

#Validate
df["DiscountPercent"].value_counts(normalize=True) * 100
```

```python
#02 — PromotionApplied
df["PromotionApplied"] = (
    df["DiscountPercent"] > 0
)
```

```python
#03 — Promotion Type
#first create the column 
df["PromotionType"] = "No Promotion"

#update only discounted sales
promotion_types = [
    "Flash Sale",
    "Seasonal Sale",
    "Clearance",
    "Member Discount"
]

mask = df["PromotionApplied"]

df.loc[
    mask,
    "PromotionType"
] = np.random.choice(
    promotion_types,
    size=mask.sum()
)

#Validate
df["PromotionType"].value_counts()
```

```python
#04 — Order Size
conditions = [
    df["SalesAmount"] < 50,
    df["SalesAmount"].between(50, 200),
    df["SalesAmount"] > 200
]

choices = [
    "Small",
    "Medium",
    "Large"
]

df["OrderSize"] = np.select(
    conditions,
    choices,
    default="Small"
)
```

```python
#05 — Sales Channel
df["SalesChannel"] = np.random.choice(
    [
        "Online",
        "In-Store"
    ],
    size=len(df),
    p=[0.60, 0.40]
)
```

```python
#06 — Payment Method
payment_methods = [
    "Credit Card",
    "Debit Card",
    "PayPal",
    "Bank Transfer",
    "Cash"
]

payment_probabilities = [
    0.50,
    0.20,
    0.15,
    0.10,
    0.05
]

df["PaymentMethod"] = np.random.choice(
    payment_methods,
    size=len(df),
    p=payment_probabilities
)
```

```python
#07 — Shipping Priority
shipping = [
    "Standard",
    "Express",
    "Same Day"
]

shipping_prob = [
    0.65,
    0.25,
    0.10
]

df["ShippingPriority"] = np.random.choice(
    shipping,
    size=len(df),
    p=shipping_prob
)
```

```python
#08 — Validation
df[
    [
        "SalesAmount",
        "DiscountPercent",
        "PromotionApplied",
        "PromotionType",
        "OrderSize",
        "SalesChannel",
        "PaymentMethod",
        "ShippingPriority"
    ]
].head(20)
```

```python
#09 — Save the Dataset
df.to_csv(
    "../data/processed/transactions_enriched.csv",
    index=False
)
```
